```python

import matplotlib.pyplot as plt
import pandas as pd

# Read the CSV file
df = pd.read_csv("stock_data.csv", parse_dates=["Date"])

# Sort data by date
df = df.sort_values("Date")

# Create figure
plt.figure(figsize=(12, 6))

# Plot closing price
plt.plot(
    df["Date"],
    df["Closing Price"],
    color="blue",
    linewidth=2,
    marker="*",
    markersize=4
)

# Add title
plt.title(
    "Closing Price Trend over Time",
    fontsize=16,
    fontweight="bold"
)

# Add axis labels
plt.xlabel("Date")
plt.ylabel("Closing Price")

# Add grid
plt.grid(True, linestyle="--", alpha=0.6)

# Rotate date labels
plt.xticks(rotation=45)

# Adjust layout
plt.tight_layout()

# Save graph
plt.savefig("line_plot.png", dpi=300)

# Display graph
plt.show()
```
